```python
import xarray as xr
import shutil
import os

#Combining datasets
directory = "/glade/derecho/scratch/considine"

datasets = ["13_14_f_30d","14_02_10_f_30d","15_16_f_30d"]

coordinates = ["output_0128_0056"]

for coordinate in coordinates:

    new_folder = f"{directory}/11_16_f_30d/{coordinate}"
    try:
        os.mkdir(f"{directory}/11_16_f_30d/{coordinate}")
    except:
        print(f"Directory exists: {directory}/11_16_f_30d/{coordinate}")

    for dataset in datasets:

        folder = f"{directory}/{dataset}/{coordinate}"
        
        for file in os.listdir(folder):

            if file == "debug":

                try:
                    os.mkdir(f"{new_folder}/debug{dataset}")
                except:
                    print("")

                for item in os.listdir(f"{folder}/debug"):
                    shutil.move(f"{folder}/debug/{item}", f"{new_folder}/debug{dataset}")
                    
            elif file.startswith("forward") == True:
                shutil.move(f"{folder}/{file}", new_folder)
```
